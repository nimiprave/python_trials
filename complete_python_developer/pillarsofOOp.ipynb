1.Encapsulation -- Binding of data and object. i.e. attributes and methods of the class.


```python
class PlayerCharacter:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def speak(self):
        print(f'My name is {self.name}, and I am {self.age} years old')
```

```python
# Instantiate the class
player1 = PlayerCharacter('Nimi', 24)
player1.speak()  # My name is Nimi, and I am 24 years old
```

Abstraction: hidding of the information and getting access only to the information that is required. The implementation details are hidden from the user.


Private vs Public Variables


```python
class Player:
    def __init__(self, name, age):
        self._name = name  # there is no private variable scope in python, instead we use _ to indicate that the variable is private
        self._age = age

    def speak(self):
        print(f'My name is {self._name}, and I am {self._age} years old')

    def getSelf(self):
        return self


ply1 = Player('Nimi', 24)
ply1.speak()  # My name is Nimi, and I am 24 years old
# This is not a good practice, but it is possible to change the private variable
ply1._name = 'John'
ply1.speak()
# can i create a new attribute from the class instance
ply1.country = 'Nigeria'
print(ply1.country)  # Nigeria

# can I access the self attribute from the class instance
print(f'{ply1.getSelf().country}')
```

Inheritence: Allows the usage of the attributes / methods of the class being inherited.


```python
# example of a class inheritence

class User:
    
    def __init__(self, email):
        self.email = email
    def sign_in(self):
        print(f'logged in user: {self.email}.')
        
        
class Arrow(User):
    def __init__(self, name, no_of_arrows, email):
        super().__init__(email)
        self.name = name
        self.no_of_arrows = no_of_arrows
        
    def attack(self):
        print(f'attacking with {self.no_of_arrows} arrows')
        

class Wizard(User):
    def __init__(self, name, power,email):
        super().__init__(email)
        self.name = name
        self.power = power
        
    def attack(self):
        print(f'attacking with power of {self.power}')
        
    
wizard1 = Wizard('Merlin', 50, 'ni@gmail.com')
arrow1 = Arrow('Robin', 100, 'syam@gmail.com')
wizard1.sign_in()
arrow1.sign_in()
wizard1.attack()
arrow1.attack()
```

Polymorphism    -- different object classes share the method names.             

```python
class Pets():
    animals = []
    def __init__(self, animals):
        self.animals = animals
    
    def walk(self):
        for animal in self.animals:
            print(animal.walk())


class Cat():
    is_lazy = True
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
    def walk(self):
        return f'{self.name} is just walking around'
    
    
class Simon(Cat):
    def sing(self, sounds):
        return f'{sounds}'
    
class Sally(Cat):
    def sing(self, sounds):
        return f'{sounds}'    
    
class Tommy(Cat):
    def sing(self, sounds):
        return f'{sounds}'
    

sally = Sally('Sally', 4)
simon = Simon('Simon',3)
tommy = Tommy('Tommy', 5)   
my_cats = [sally, simon, tommy]

my_pets = Pets(my_cats)
my_pets.walk()  # Sally is just walking around
                # Simon is just walking around

                
```

Decorators: 
Demonstration of the usage of Decorators. Decators work on the concept of high order functions.

What are highorder functions?
 High Order functions have the capability to accept function as a parameter and can also return a function. 


```python
# high order function example

def greet():
    print('Hello')


def showGreeting(greet):
    greet()


showGreeting(greet)
showGreeting(lambda: print('I am lambda function'))

```

```python
#higher order function returning a function
def greetingFactory():
    def greetingInstance():
        print('This is a greeting instance')
    
    return greetingInstance


greetingFactory()()
```

Decarators Example: Decarator pattern should typically take one function as parameter , has an inside functions that can decorate the given function and return the wrapper function. The decorator can be used by using the @<Decorator name>.     

```python
def decorate(greetings):
    def wrapper():
        print('************************')
        greetings()
        print('************************')
        
    return wrapper

@decorate
def sayHello():
    print('Hey There!.. How are you')
    
sayHello()
```

```python
# Create an @authenticated decorator that only allows the function to run is user1 has 'valid' set to True:
user1 = {
    'name': 'Sorna',
    'valid': True 
}

user2 = {
    'name': 'NImi',
    'valid': False 
}


def authenticated(fn):
    def wrapper(user):
        if user['valid'] == True:
            print(f'User is valid to execute the function')
            fn(user)
        else:
            print(f'{user['name']} is not valid to run the function ')
            
    return wrapper

@authenticated
def message_friends(user):
    print('Message has been sent')        

    
message_friends(user1)   
message_friends(user2)   
            
```

Demonstration of the usage of Functions in Python.
Functions address seperation of concerns and is basically used to isolate the data and the function. The function will perform an action on the data provided and does not have arbitrary relations with the data itself.

Types of functions:
1.Map
2.filter
3.zip
4.reduce

```python
#function Map demonstration. Map will perform some action on the elements of the iterable

my_list = [1,2,3,4,5,6]

def square(item):
    return item * item

squared_list = map(square,my_list)
print(squared_list)

for item in squared_list:
    print(f'The squared values are {item}')



```

Filter Function:  filters the data provided to the function and return boolean based on the condition the function is implementing.

```python
my_newlist = [1,2,3,4,5,6]

def even(item):
    return item % 2 == 0


print(my_newlist.__str__())
print(even(3))
print(even(4))
even_filtered = filter(even,my_newlist)
for item in even_filtered:
    print(item)

```

zip() function demonstration    -- adding the two iterables together, element by element, the size of the two iterables should be the same otherwise, the other elements from the second list will be ignored. 

```python
first_list = [1,2,3,4,]
second_list = [5,6,7,8,9,10]
third_list = ['a','b','c','d','e','f']

result = zip(third_list, first_list, second_list)
print(result)
for item in result:
    print(item)
    
print('done')


```

Reduce() -- function for reducing the iterators to some value . (i.e.) given a list, we can find the sum of the list, average of the list etc. we need to import the fuction from the functools module.

```python
#Reduce Demo
from functools import reduce
mylist = [1,2,3,4,5]

def totalsum(acc, item):
    return acc + item

def average(list):
   total = reduce(totalsum,list)
   return total / len( list )

print(reduce(totalsum,mylist))
print(average(mylist))
```

Bad data could be:

Empty cells
Data in wrong format
Wrong data
Duplicates

Removing the Empty Rows.

```python
import pandas as pd

#read off the csv file
df = pd.read_csv('dataset.csv')
#print('Before Cleaning the Empty values')
#print(df)

new_df = df.dropna()
print('After Cleaning the Empty values')
print(new_df.to_string())
```

Replace Empty Values with some new values

```python
import pandas as pd
df = pd.read_csv('dataset.csv')
#print(df.to_string())
df.fillna(130, inplace = True)
print(df.to_string())
```

Replacing the Empty Values with the Mean, Median , Mode. A common way to replace empty cells, is to calculate the mean, median or mode value of the column. Pandas uses the mean() median() and mode() methods to calculate the respective values for a specified column:

```python
import pandas as pd
df = pd.read_csv('dataset.csv')
#print(df.to_string())
calories_median = df['Calories'].median()
print(f'Median of Calories: {calories_median}')
df['Calories'].fillna(calories_median, inplace = True)
print(df.to_string())
```

Formatting the date of the Column

```python
import pandas as pd
df = pd.read_csv('dataset.csv')
df['Date'] = pd.to_datetime(df['Date'],format='mixed')
print(df.to_string())
```

```python
-Fill the empty columns of the calories with the median of the column.
-Fill the empty field of the date column with the most recent date in the column.
-format the wrong date format in the date column. 
```

```python
import pandas as pd
from datetime import datetime

df = pd.read_csv('dataset.csv')
formatted_date = datetime.today().date()
print(formatted_date)
#Replace all empty cells with the median in the calories column
df['Calories'].fillna(df['Calories'].median(), inplace=True)
df['Date'].fillna(formatted_date,inplace=True)
df['Date'] = pd.to_datetime(df['Date'],format='mixed')
print(df)

```

Replacing the values row by row. 

```python
import pandas as pd
from datetime import datetime
df = pd.read_csv('dataset.csv')

#check if the duration is greater than 60 mins, if so bound to 60 mins. 

for x in df.index:
   # print(x)
    if (df.loc[x, "Duration"]) > 60:
            df.loc[x,"Duration"] = 60
    if pd.isna(df.loc[x,"Calories"]):
        df.loc[x,"Calories"] = df["Calories"].mean()
    if pd.isna(df.loc[x,"Date"]):
        df.loc[x,"Date"] = pd.to_datetime(datetime.today().date(),format='mixed') 

df['Date'] = pd.to_datetime(df['Date'],format='mixed')
print(df.to_string())


```

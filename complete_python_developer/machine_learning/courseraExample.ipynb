```python
def convert_to_currency(value):
    word = value.split('€')[1]
    if word[-1] == 'M':
        return float(word[:-1])*1000000
    else:
        return float(word[:-1])*1000
    
for wage in three_columns['Wage']:
     print(convert_to_currency(wage))    
```
